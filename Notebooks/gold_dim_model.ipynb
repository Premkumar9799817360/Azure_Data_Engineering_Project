## CREATE DIMENSTION MODEL

### Fetch Relative Columns

```python
%sql
select distinct(Model_ID)as Model_ID,model_category from parquet.`abfss://silver@projectdatastorages.dfs.core.windows.net/carsales`
```

```python
#create datafram  above sql 
df_src = spark.sql(
    '''
    select distinct(Model_ID)as Model_ID,model_category from parquet.`abfss://silver@projectdatastorages.dfs.core.windows.net/carsales`
    '''
)
```

```python
df_src.display()
```

```python
#create key 
df_sink = spark.sql(
    '''
    SELECT 1 AS dim_model_key, Model_ID,model_category
    from parquet.`abfss://silver@projectdatastorages.dfs.core.windows.net/carsales`
    WHERE 1 = 0
    '''
)
```

```python
df_sink.display()
```

### dim_model Sink - Initial and Incremental

```python
# else work as first time create table and if work as incremental load data second time store data

if spark.catalog.tableExists("car_catalog.gold.dim_model"):

  df_sink = spark.sql(
    '''
    SELECT dim_model_key, Model_ID,model_category
    from car_catalog.gold.dim_model
    '''
  )

else:
  df_sink = spark.sql(
    '''
    SELECT 1 AS dim_model_key, Model_ID,model_category
    from parquet.`abfss://silver@projectdatastorages.dfs.core.windows.net/carsales`
    WHERE 1 = 0
    '''
  )
```

```python
df_sink.display()
```

### Filtering new records and old records

```python
df_filter = df_src.join(df_sink,df_src['Model_ID'] == df_sink['Model_ID'],'left').select(df_src['Model_ID'],df_src['model_category'],df_sink['dim_model_key'])
df_filter.display()
```

#### df_filter_old

```python
from pyspark.sql.functions import *
from pyspark.sql.types import  *
df_filter_old = df_filter.filter(col("dim_model_key").isNotNull())
```

```python
df_filter_old.display()
```

#### df_filter_new

```python
df_filter_new = df_filter.filter(col("dim_model_key").isNull()).select("Model_ID","model_category")
df_filter_new.display()
```

#### CREATE SURROGATE KEY

**Fetch the max surrogate key from existing table**

```python
if incremental_flag == '0':
    max_value = 1
else:
    max_value_df = spark.sql("select max(dim_model_key) from car_catalog.gold.dim_model")
    max_value = max_value_df.collect()[0][0]+1
    print(max_value)
```

**Create Surrogate key columns and Add max Surrogate key**

```python
df_filter_new = df_filter_new.withColumn("dim_model_key",max_value+monotonically_increasing_id())
```

```python
df_filter_new.display()
```

### Create Final DF = df_filter_old + df_filter_new

```python
df_final= df_filter_new.union(df_filter_old)
```

```python
df_final.display()
```

#SCD TYPE -1  (UPSERT)

```python
from delta.tables import DeltaTable
```

```python
from delta.tables import DeltaTable

# Incremental RUN
if spark.catalog.tableExists("car_catalog.gold.dim_model"):
    delta_tbl = DeltaTable.forPath(
        spark,
        "abfss://gold@projectdatastorages.dfs.core.windows.net/dim_model"
    )
    delta_tbl.alias("trg").merge(
        df_final.alias("src"),
        "trg.dim_model_key = src.dim_model_key"
    ).whenMatchedUpdateAll().whenNotMatchedInsertAll().execute()
    print("Merge completed")

# Initial RUN
else:
    df_final.write.format('delta').mode('overwrite').option(
        "path",
        "abfss://gold@projectdatastorages.dfs.core.windows.net/dim_model"
    ).saveAsTable("car_catalog.gold.dim_model")
```
